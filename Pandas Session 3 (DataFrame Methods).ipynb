##### value_counts() (Series and DataFrames) --> returns the frequency of a row

```python

#find who won the most number of player of matches in qualifiers and finals
matches=ipl[~ipl['MatchNumber'].str.isdigit()]
matches['Player_of_Match'].value_counts()
```

```python
#toss decision plot
ipl['TossDecision'].value_counts().plot(kind='pie')
```

```python
#how many matches each team has played

(ipl['Team1'].value_counts()+ipl['Team2'].value_counts()).sort_values()
```

##### sort_values() (Series and DataFrames) --> sorts the values in columns

```python
x = pd.Series([12,14,1,56,89])
x
```

```python
# reset_index() --> resets the indices which got scrambled during sorting process (drop=True --> drops the scrambled indices)
x.sort_values().reset_index(drop=True)
```

```python
movies.sort_values(['year_of_release','title_x'],ascending=[False,True])
```

##### rank() (Series)

```python
# rank (series)
batsman = pd.read_csv('~/Machine Learning/Pandas/Datasets/batsman_runs_ipl.csv')
batsman.head()
```

```python
batsman_rank=batsman['batsman_run'].rank(ascending=False).sort_values()
batsman['rank']=batsman_rank
batsman
```
