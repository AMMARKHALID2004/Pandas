### Concatenation

```python
import numpy as np
import pandas as pd
```

```python
students=pd.read_csv("~/Machine Learning/Pandas/Datasets/students.csv")
courses=pd.read_csv("~/Machine Learning/Pandas/Datasets/courses.csv")
nov=pd.read_csv("~/Machine Learning/Pandas/Datasets/reg-month1.csv")
dec=pd.read_csv("~/Machine Learning/Pandas/Datasets/reg-month2.csv")
ipl=pd.read_csv("~/Machine Learning/Pandas/Datasets/ipl-matches.csv")
deliveries=pd.read_csv("~/Machine Learning/Pandas/Datasets/deliveries.csv")
```

##### concat() --> used for concatenating two tables

```python
nov
```

```python
dec
```

```python
# ignore_index=True will make sure that your original indices are not repeated but rather new indices are made
pd.concat([nov,dec],ignore_index=True)
```

```python
# if we want to make multiple indices we can use keys parameter
pd.concat([nov,dec],keys=['Nov','Dec'])
```

```python
#concatenating horizontally 
pd.concat([nov,dec],axis=1)
```

### Merge using Joins

```python
regs=pd.concat([nov,dec],ignore_index=True)
regs
```

```python
#joins (inner,outer,left,right,self) --> same like SQL
students.merge(regs,how='inner',on='student_id')
```

##### 1. find total revenue generated

```python
courses.merge(regs,on='course_id')['price'].sum()
```

##### 2. find month by month revenue

```python
both_months=pd.concat([nov,dec],keys=['Nov','Dec']).reset_index()
both_months.merge(regs,on='student_id')
```
